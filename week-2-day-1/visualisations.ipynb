```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

df = sns.load_dataset("tips")

df.head()
```

```python
# Line Chart: Average Total Bill by Day

daily_sales = df.groupby("day", observed=True)["total_bill"].mean()

plt.figure(figsize=(8, 5))
plt.plot(daily_sales.index, daily_sales.values, marker="o")

plt.title("Average Total Bill by Day")
plt.xlabel("Day")
plt.ylabel("Average Total Bill ($)")

# Data callout
max_day = daily_sales.idxmax()
max_value = daily_sales.max()

plt.annotate(
    f"Highest: ${max_value:.2f}",
    xy=(max_day, max_value),
    xytext=(10, 10),
    textcoords="offset points",
    arrowprops=dict(arrowstyle="->")
)

plt.tight_layout()
plt.savefig("week-2-day-1/line_chart.png", dpi=300)
plt.show()
```

```python
# Line Chart: Average Total Bill by Day

daily_sales = df.groupby("day", observed=True)["total_bill"].mean()

plt.figure(figsize=(8, 5))
plt.plot(daily_sales.index, daily_sales.values, marker="o")

plt.title("Average Total Bill by Day")
plt.xlabel("Day")
plt.ylabel("Average Total Bill ($)")

# Data callout
max_day = daily_sales.idxmax()
max_value = daily_sales.max()

plt.annotate(
    f"Highest: ${max_value:.2f}",
    xy=(max_day, max_value),
    xytext=(10, 10),
    textcoords="offset points",
    arrowprops=dict(arrowstyle="->")
)

plt.tight_layout()
plt.savefig("week-2-day-1/line_chart.png", dpi=300)
plt.show()
```

```python
# Bar Chart: Average Tip by Day

avg_tip = df.groupby("day", observed=True)["tip"].mean()

plt.figure(figsize=(8, 5))
bars = plt.bar(avg_tip.index, avg_tip.values)

plt.title("Average Tip by Day")
plt.xlabel("Day")
plt.ylabel("Average Tip ($)")

# Data callouts
for bar, value in zip(bars, avg_tip.values):
    plt.annotate(
        f"${value:.2f}",
        xy=(bar.get_x() + bar.get_width() / 2, value),
        xytext=(0, 5),
        textcoords="offset points",
        ha="center"
    )

plt.tight_layout()
plt.savefig("week-2-day-1/bar_chart.png", dpi=300)
plt.show()
```

```python
# Histogram: Distribution of Total Bills

plt.figure(figsize=(8, 5))

plt.hist(
    df["total_bill"],
    bins=15,
    edgecolor="black"
)

plt.title("Distribution of Total Bills")
plt.xlabel("Total Bill ($)")
plt.ylabel("Number of Customers")

plt.tight_layout()
plt.savefig("week-2-day-1/histogram.png", dpi=300)
plt.show()
```

```python
# Box Plot: Total Bill by Day

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="day",
    y="total_bill"
)

plt.title("Distribution of Total Bill by Day")
plt.xlabel("Day")
plt.ylabel("Total Bill ($)")

plt.tight_layout()
plt.savefig("week-2-day-1/box_plot.png", dpi=300)
plt.show()
```
