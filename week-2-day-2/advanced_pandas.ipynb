```python
import pandas as pd
import numpy as np

print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
```

```python
customers = pd.DataFrame({
    "customer_id": [101, 102, 103, 104, 105],
    "customer_name": ["Aarav", "Priya", "Rahul", "Sneha", "Vikram"],
    "region": ["South", "West", "North", "South", "East"]
})

orders = pd.DataFrame({
    "order_id": [1001, 1002, 1003, 1004, 1005, 1006],
    "customer_id": [101, 102, 103, 101, 104, 106],
    "product": ["Laptop", "Phone", "Tablet", "Mouse", "Keyboard", "Monitor"],
    "sales": [55000, 25000, 18000, 1500, 3000, 12000]
})

print("Customers:")
display(customers)

print("\nOrders:")
display(orders)
```

```python
# INNER JOIN
inner_join = pd.merge(
    customers,
    orders,
    on="customer_id",
    how="inner"
)

print("INNER JOIN")
display(inner_join)
```

```python
# LEFT JOIN
left_join = pd.merge(
    customers,
    orders,
    on="customer_id",
    how="left"
)

print("LEFT JOIN")
display(left_join)
```

```python
# RIGHT JOIN
right_join = pd.merge(
    customers,
    orders,
    on="customer_id",
    how="right"
)

print("RIGHT JOIN")
display(right_join)
```

```python
# OUTER JOIN
outer_join = pd.merge(
    customers,
    orders,
    on="customer_id",
    how="outer"
)

print("OUTER JOIN")
display(outer_join)
```

```python
# Sales data for pivot table

sales_data = pd.DataFrame({
    "region": [
        "South", "South", "South",
        "North", "North", "North",
        "West", "West", "West",
        "East", "East", "East"
    ],
    "product": [
        "Laptop", "Phone", "Laptop",
        "Phone", "Tablet", "Laptop",
        "Laptop", "Phone", "Tablet",
        "Phone", "Tablet", "Laptop"
    ],
    "sales": [
        55000, 25000, 60000,
        30000, 18000, 52000,
        48000, 22000, 15000,
        27000, 20000, 45000
    ]
})

display(sales_data)
```

```python
# Pivot table: total sales by region and product

pivot_sales = sales_data.pivot_table(
    index="region",
    columns="product",
    values="sales",
    aggfunc="sum",
    fill_value=0
)

print("Sales by Region and Product")
display(pivot_sales)
```

```python
# Create a wide-format DataFrame

wide_sales = pd.DataFrame({
    "region": ["South", "North", "West", "East"],
    "Laptop": [115000, 52000, 48000, 45000],
    "Phone": [25000, 30000, 22000, 27000],
    "Tablet": [0, 18000, 15000, 20000]
})

print("Wide format:")
display(wide_sales)
```

```python
# Reshape wide data into long format

long_sales = pd.melt(
    wide_sales,
    id_vars=["region"],
    var_name="product",
    value_name="sales"
)

print("Long format:")
display(long_sales)
```

```python
# Method chaining: filter, group, aggregate, sort, and rename

summary = (
    sales_data
    .query("sales >= 20000")
    .groupby(["region", "product"], as_index=False)
    .agg(total_sales=("sales", "sum"))
    .sort_values("total_sales", ascending=False)
    .reset_index(drop=True)
)

print("Sales Summary using Method Chaining:")
display(summary)
```
