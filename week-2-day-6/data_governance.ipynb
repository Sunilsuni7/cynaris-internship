```python
# Recalculate quality checks for the scorecard

# Accuracy
age_invalid = df[
    (df["age"].notna()) &
    ((df["age"] < 0) | (df["age"] > 100))
]

fare_invalid = df[
    (df["fare"].notna()) &
    (df["fare"] < 0)
]

# Consistency
invalid_sibsp = df[df["sibsp"] < 0]
invalid_parch = df[df["parch"] < 0]
invalid_pclass = df[~df["pclass"].isin([1, 2, 3])]

# Uniqueness
duplicate_count = df.duplicated().sum()

# Validity
valid_sex = {"male", "female"}
valid_embarked = {"C", "Q", "S"}
valid_pclass = {1, 2, 3}

invalid_sex = df[
    df["sex"].notna() &
    ~df["sex"].isin(valid_sex)
]

invalid_embarked = df[
    df["embarked"].notna() &
    ~df["embarked"].isin(valid_embarked)
]

invalid_pclass_validity = df[
    ~df["pclass"].isin(valid_pclass)
]

print("Quality check variables recalculated successfully.")
print("Invalid age:", len(age_invalid))
print("Invalid fare:", len(fare_invalid))
print("Duplicate records:", duplicate_count)
print("Invalid sex:", len(invalid_sex))
print("Invalid embarked:", len(invalid_embarked))
```
