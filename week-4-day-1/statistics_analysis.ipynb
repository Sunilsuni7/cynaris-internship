```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    "Group": [
        "A", "A", "A", "A", "A",
        "B", "B", "B", "B", "B"
    ],
    "Sales": [
        120, 135, 128, 142, 125,
        180, 195, 210, 175, 200
    ]
}

df = pd.DataFrame(data)

df
```

```python
mean = df["Sales"].mean()
median = df["Sales"].median()
mode = df["Sales"].mode()[0]
variance = df["Sales"].var()
std_dev = df["Sales"].std()

print("Mean:", mean)
print("Median:", median)
print("Mode:", mode)
print("Variance:", variance)
print("Standard Deviation:", std_dev)
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(df["Sales"], bins=5, kde=True)

plt.title("Sales Frequency Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
q1 = df["Sales"].quantile(0.25)
q2 = df["Sales"].quantile(0.50)
q3 = df["Sales"].quantile(0.75)

print("25th Percentile (Q1):", q1)
print("50th Percentile (Q2 / Median):", q2)
print("75th Percentile (Q3):", q3)
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x="Group", y="Sales")

plt.title("Sales Distribution by Group")
plt.xlabel("Group")
plt.ylabel("Sales")
plt.show()
```

```python
print("Interpretation:")
print("- Mean represents the average sales value.")
print("- Median represents the middle sales value.")
print("- Mode represents the most frequently occurring sales value.")
print("- Variance and standard deviation describe the spread of sales.")
print("- Q1, Q2, and Q3 describe the distribution using quartiles.")
print("- The box plot helps compare the sales distributions of Groups A and B.")
```
